```python
# ============================================
# CELLA 6: Scarica i video .avi dalla cartella output_videos
# ============================================
from google.colab import files
import os
from pathlib import Path

# Accedi alla directory output_videos
output_dir = 'output_videos'

if os.path.exists(output_dir):
    # Trova tutti i file .avi nella directory
    avi_files = list(Path(output_dir).glob('*.avi'))

    if avi_files:
        print(f"✅ Trovati {len(avi_files)} file .avi da scaricare:\n")
        for file in avi_files:
            print(f"  📥 Scaricamento: {file.name}")
            files.download(str(file))
        print(f"\n✅ Scaricamento completato!")
    else:
        print(f"❌ Nessun file .avi trovato in {output_dir}")
        print(f"File presenti: {os.listdir(output_dir)}")
else:
    print(f"❌ Directory {output_dir} non trovata")
    print("Directory presenti:", os.listdir())
```
