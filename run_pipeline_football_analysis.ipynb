# Football Analysis — Full Pipeline

This notebook runs the complete football match analysis pipeline:
- Player, referee, and ball detection & tracking (YOLOv8 + ByteTrack)
- Camera movement and zoom compensation
- Dynamic field area detection
- Real-world position mapping (pixels → metres)
- Speed & distance estimation
- Team assignment by jersey colour (K-Means)
- Ball possession
- **Offside detection**

```python
# ============================================
# CELL 7: Download output video
# ============================================
from google.colab import files
from pathlib import Path
import os

output_dir = 'output_videos'

if os.path.exists(output_dir):
    avi_files = list(Path(output_dir).glob('*.avi'))

    if avi_files:
        print(f"Found {len(avi_files)} output video(s):\n")
        for file in avi_files:
            print(f"  Downloading: {file.name}")
            files.download(str(file))
        print("\nDownload complete!")
    else:
        print(f"No .avi files found in {output_dir}")
        if os.path.exists(output_dir):
            print(f"Contents: {os.listdir(output_dir)}")
else:
    print(f"Directory {output_dir} not found.")
    print("Available directories:", [d for d in os.listdir() if os.path.isdir(d)])
```
